```
import matplotlib.pyplot as plt
import numpy as np

# Загрузка данных
data = np.genfromtxt("wine.data.txt", delimiter=",", dtype=None, encoding=None)

# Разделяем данные по классам
class1 = data[data['f0'] == 1]
class2 = data[data['f0'] == 2]
class3 = data[data['f0'] == 3]

# Проекция 1: Алкоголь — Яблочная кислота
plt.figure(1)
plt.plot(class1['f1'], class1['f2'], 'ro', label='Класс 1')
plt.plot(class2['f1'], class2['f2'], 'g^', label='Класс 2')
plt.plot(class3['f1'], class3['f2'], 'bs', label='Класс 3')
plt.xlabel('Алкоголь')
plt.ylabel('Яблочная кислота')
plt.legend(bbox_to_anchor=(1.05, 1), loc=2)
plt.title('Проекция 1: Алкоголь — Яблочная кислота')
plt.show()

# Проекция 2: Алкоголь — Общие фенолы
plt.figure(2)
plt.plot(class1['f1'], class1['f6'], 'ro', label='Класс 1')
plt.plot(class2['f1'], class2['f6'], 'g^', label='Класс 2')
plt.plot(class3['f1'], class3['f6'], 'bs', label='Класс 3')
plt.xlabel('Алкоголь')
plt.ylabel('Общие фенолы')
plt.legend(bbox_to_anchor=(1.05, 1), loc=2)
plt.title('Проекция 2: Алкоголь — Общие фенолы')
plt.show()

# Проекция 3: Интенсивность цвета — Оттенок
plt.figure(3)
plt.plot(class1['f10'], class1['f11'], 'ro', label='Класс 1')
plt.plot(class2['f10'], class2['f11'], 'g^', label='Класс 2')
plt.plot(class3['f10'], class3['f11'], 'bs', label='Класс 3')
plt.xlabel('Интенсивность цвета')
plt.ylabel('Оттенок')
plt.legend(bbox_to_anchor=(1.05, 1), loc=2)
plt.title('Проекция 3: Интенсивность цвета — Оттенок')
plt.show()
```
